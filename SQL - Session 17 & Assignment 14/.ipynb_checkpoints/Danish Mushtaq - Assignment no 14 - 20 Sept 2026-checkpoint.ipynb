```python
import pandas as pd
import sqlite3 as sql
```

```python
conn = sql.connect('database.sqlite')
conn
```

```python
def sql_query (q):
    return pd.read_sql_query(q,conn)
```

```python
sql_query('select *from salaries')
```

```python
q = 'select *from salaries'
sql_query (q)
```

```python
q = 'pragma table_info (salaries)'
sql_query(q)
```

#### **1. What is the average TotalPayBenefits for each JobTitle? exclude ('Not Provided', 'Not provided')**

```python
q = """Select jobtitle, avg(totalpaybenefits) as averagetotalpaybenefits from salaries where jobtitle not in ('Not Provided', 'Not provided')
group by jobtitle order by averagetotalpaybenefits desc;"""
sql_query (q)
```

#### **2. Show The JobTitle and Overtime Pay for All Employees With Overtime Pay greater than 50000**

```python
q = """Select jobtitle,overtimepay from salaries where overtimepay > 50000;"""
sql_query (q)
```

#### **3. What is the total number of unique employees? exclude ('Not Provided', 'Not provided')**

```python
q = """Select count(distinct id) as uniqueemployees from salaries where id not in ('not provided', 'not provided');"""
sql_query(q)
```

#### **4.Show The Top 10 highest Paid Employees**

```python
q = """select employeename, jobtitle, totalpaybenefits from salaries order by totalpaybenefits desc limit 10"""
sql_query(q)

```

#### **5.Show All Employees Ordered By Their TotalPay Benefits In Descending Order**

```python
q = """Select EmployeeName, JobTitle, TotalPayBenefits From Salaries order by TotalPayBenefits desc;"""
sql_query (q)
```
