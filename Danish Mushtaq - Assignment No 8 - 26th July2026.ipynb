Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, Name, Number, Address, Salary, Post, Education, Year_of_Joining, Year_of_Experience):
        self.Name = Name
        self.Number = Number
        self.Address = Address
        self.Salary = Salary
        self.Post = Post
        self.Education = Education
        self.Year_of_Joining = Year_of_Joining
        self.Year_of_Experience = Year_of_Experience

    
    def describe_employee (self):
        print("__Employee Information__")
        print("Name", self.Name)
        print("Employee Number", self.Number)
        print("Address", self.Address)
        print("Salary", self.Salary)
        print("Post", self.Post)
        print("Education", self.Education)
        print("Year of Joining", self.Year_of_Joining)
        print("Year of Experience", self.Year_of_Experience)

    def update_year_of_experience (self, Year):
        self.Year_of_Experience += Year
        print(f"{self.Name} exprience update to {self.Year_of_Experience} year")

    def Salary_Increase (self, Amount):
        self.Salary += Amount
        print(f"{self.Name} new slary is {self.Salary}")
```

```python
employee420 = Employee("Danish",420,"Model Colony",50000,"Data Analyst","M.com",2021,4)

print("Name:", employee420.Name)
print("Number:", employee420.Number)
print("Salary:", employee420.Salary)
print("Post:", employee420.Post)

employee420.describe_employee()
employee420.Salary_Increase(500)
employee420.update_year_of_experience(4)
```

#### Question 2

Complete the same task (Question 1) without using the Classes or OOP.

```python
def create_employee(name, number, address, salary, post, year_of_joining, education, year_of_experience):
    return {"name": name,"number": number,"address": address,"salary": salary,"post": post,"year_of_joining": year_of_joining,"education": education,
    "year_of_experience": year_of_experience}

def describe_employee(employee):
    print("__Employee Information__")
    print("Name:", employee["name"])
    print("Employee Number:", employee["number"])
    print("Address:", employee["address"])
    print("Salary:", employee["salary"])
    print("Post:", employee["post"])
    print("Year of Joining:", employee["year_of_joining"])
    print("Education:", employee["education"])
    print("Years of Experience:", employee["year_of_experience"])

def Salary_Increase(employee, amount):
    employee["salary"] += amount
    print(employee["name"], "'s new salary is", employee["salary"])

def update_year_of_experience(employee, years):
    employee["year_of_experience"] += years
    print(employee["name"], "'s experience is now", employee["year_of_experience"], "years")
```

```python
employee420 = create_employee("Danish",420,"Model Colony",50000,"Data Analyst", 2021 , "MBA", 4)

print("Name:", employee420["name"])
print("Employee Number:", employee420["number"])
print("Salary:", employee420["salary"])
print("Post:", employee420["post"])

describe_employee(employee420)
Salary_Increase(employee420, 500)
update_year_of_experience(employee420, 4)
```
