1. List all cities where the area is greater than 5000, along with their population.

```python
q = 'Select *from cities where population > 5000'
sql_query (q)
```

2. List the names of capital cities where the death rate is greater than 2% and the
birth rate is less than 20%

```python
q = '''
SELECT cities.name
FROM facts
INNER JOIN cities
    ON facts.id = cities.facts_id
WHERE cities.capital = 1
  AND facts.death_rate > 2
  AND facts.birth_rate < 20
'''

sql_query(q)
```

```python
q = '''
SELECT cities.name,
       facts.death_rate,
       facts.birth_rate
FROM facts
INNER JOIN cities
    ON facts.id = cities.facts_id
WHERE cities.capital = 1
  AND facts.death_rate > 2
  AND facts.birth_rate < 20
'''

sql_query(q)
```

```python
q = '''
SELECT facts.name AS country,
       cities.name AS capital_city,
       facts.death_rate,
       facts.birth_rate
FROM facts
INNER JOIN cities
    ON facts.id = cities.facts_id
WHERE cities.capital = 1
  AND facts.death_rate > 2
  AND facts.birth_rate < 20
'''

sql_query(q)
```

```python
q = '''
SELECT facts.name AS country,
       cities.name AS capital_city,
       facts.death_rate,
       facts.birth_rate
FROM facts
INNER JOIN cities
    ON facts.id = cities.facts_id
WHERE facts.name = 'Pakistan'
  AND cities.name = 'Islamabad'
'''

sql_query(q)
```

3. List the cities with a population growth rate between 1% and 5%, sorted in desc
by population growth.

```python
q = '''select name, population_growth from facts 
where population_growth between 1 and 5
order by population_growth desc'''
sql_query(q)
```

```python
q = '''select facts.name AS country, name, population_growth from facts 
where population_growth between 1 and 5
order by population_growth desc'''
sql_query(q)
```

```python
q = '''
SELECT facts.name AS country,
       cities.name AS city,
       facts.population_growth
FROM facts
INNER JOIN cities
    ON facts.id = cities.facts_id
WHERE facts.name = 'Pakistan'
  AND cities.name = 'Islamabad'
  AND facts.population_growth BETWEEN 1 AND 5
ORDER BY facts.population_growth DESC
'''

sql_query(q)
```

##### 4. Find countries that start with &quot;A&quot; and do not have any cities listed in the cities table. The query should return the country name and city name. Using left join.

```python
q = '''
select facts.name as country, cities.name as city
from facts left join cities ON facts.id = cities.facts_id
WHERE facts.name LIKE 'A%' AND cities.name IS NULL'''
sql_query(q)
```

```python
q = '''
SELECT facts.name AS country,
       cities.name AS city
FROM facts
LEFT JOIN cities
    ON facts.id = cities.facts_id
WHERE facts.name LIKE 'A%'
'''

sql_query(q)
```

```python
q = '''
SELECT facts.name AS country,
       cities.name AS city
FROM facts
LEFT JOIN cities
    ON facts.id = cities.facts_id
WHERE cities.name IS NULL
'''

sql_query(q)
```

5. List all capital cities and their corresponding death rate using a LEFT JOIN.

```python
q = '''Select cities.name AS capital_city, facts.death_rate
FROM cities LEFT JOIN facts ON cities.facts_id = facts.id
WHERE cities.capital = 1 '''
sql_query(q)
```
